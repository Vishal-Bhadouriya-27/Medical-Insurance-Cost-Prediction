* ***Outliers -->***

 Extremely high or extremely low values

```python
# It is used to visualise the outliers through boxplot -->
for i in df.columns:
  if df[i].dtype != 'object':
    plt.boxplot(df[i])
    plt.xlabel(i)
    plt.show()
```

```python
# Handling Outliers using IQR (Interquartile Range).
#IQR method for mathematical treatment of outliers.

col = ['bmi' , 'past_consultations','Hospital_expenditure','NUmber_of_past_hospitalizations','Anual_Salary']
for i in col :
  q1 = df[i].quantile(0.25)
  q3 = df[i].quantile(0.75)
  IQR = q3 - q1
  LB = q1 -1.5 * IQR
  UB = q3 + 1.5 *IQR
  df = df[(df[i] >= LB) &(df[i] <= UB)]
```

```python
# To visualise the outlier after IQR implementation
for i in df.columns:
  if df[i].dtypes != 'object':
    plt.boxplot(df[i])
    plt.xlabel(i)
    plt.show()
```

* ***Perform Label Encoding in the dataset*** -->

* Encoding is used to convert the object data into other datatypes like (int, float)

```python
# I import the LabelEncoder
# Applying Encoding to all the object data
from sklearn.preprocessing import LabelEncoder
cat_cols = df.select_dtypes(include = 'object')  # Cat = categorical data. & le = labelEncoder
le = LabelEncoder()
for i in cat_cols.columns:
  df[i] = le.fit_transform(df[i])
```

* ***VIF - Variance Inflation factor .***
* VIF < 6 is good to go .
* VIF > 6 is not good


```python
#Import the variance Inflation factor
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
# Create a list of all the numberical columns.
col_list = []
for  i in df.columns:
  if (i != 'charges'):
    col_list.append(i)

# Apply VIF on columns present inside the col_list.
X = df[col_list]   # Store all the columns in the variable x
vif_data = pd.DataFrame()
vif_data ['Column_Name'] = X.columns
vif_data['VIF_score'] = [variance_inflation_factor(X.values , i) for i in range (len(X.columns))]
vif_data
```
